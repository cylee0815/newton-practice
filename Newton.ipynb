```python
def first_derivative(f, x, eps=0.00001):
    return (f(x + eps) - f(x - eps)) / 2 / eps


def second_derivative(f, x, eps=0.00001):
    return (f(x + eps) - 2 * f(x) + f(x - eps)) / eps**2
```

```python
def my_newton_method(x0, f, thres=0.001, max_count=100000):
    # for robustness
    count = 0
    x = x0 + 100000 * thres  # to run the while loop
    while abs(x - x0) > thres and count < max_count:
        x, x0 = x0 - first_derivative(f, x0) / second_derivative(f, x0), x
        count += 1
    if abs(x - x0) <= thres:
        return x
    else:
        raise ValueError(
            "Max count of {} is reached and it still can't converge".format(max_count)
        )
```
